## SETS : They are built in data structures used to store collection of unique items . They are unordered means elements do not follow a specific order , duplicate elements are not allowed , they are used to remove duplicate entries.

```python
# Create a set 

set_1 = {1,2,3,4}
print(type(set_1))
print(set_1)
```

```python
my_empty_set = set()
print(my_empty_set)
```

```python
## Conversion of an list into set

my_set = set([1,2,3,4])
print(type(my_set))
```

```python
## Even if we give duplicate elements in a list and then convert it into a set it will give only the first occurence

my_set1 = set([1,2,3,4,5,6,5,4,5,6])
print(type(my_set1))
print(my_set1)
```

```python
## Basic Set Operations

# Adding and Removing Elements

my_set1.add(7)
print(my_set1)
```

```python
my_set1.remove(5)
print(my_set1)
```

```python
my_set1.remove(11) # It will give error

my_set1.discard(11) # Remooves the element if it is in the set
```

```python
## Pop Method : It will remove the first element of the set

removed_my_set1 = my_set1.pop()

print(removed_my_set1)
print(my_set1)
```

```python
## clear all the elements

my_set1.clear()
print(my_set1)
```

```python
## Set Membership Test : Wether an element is in the set or not

my_set2 = {1,2,3,4,5,6}
print(8 in my_set2)
print(6 in my_set2)
```

```python
## Mathemaatical Operation

set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

## Union 
union_set = set1.union(set2)
print(union_set)

## Intersection
intersection_set = set1.intersection(set2)
print(intersection_set)

# Here the common elements between both sets are taken and the elements are updated in set1. 

set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

## Difference
print(set1.difference(set2))
print(set2.difference(set1))
```

```python
## Symmetric Differnce : Common elements are removed and unique elements are combined.

print(set1.symmetric_difference(set2))
```

```python
## Sets Method 
set1 = {1,2,3}
set2 = {3,4,5}

print(set1.issubset(set2))
print(set1.issuperset(set2))

## There are many more methods like this in python
```

```python
## Counting unique words in text

text = "In this tutorial we are discussing about sets"

words = text.split()
unique_words = set(words)
print(unique_words)
print(len(unique_words))
```
