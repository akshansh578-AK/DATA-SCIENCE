## Dictionaries : They are unordered collection of items . They store data in key-value pairs . Keys are unique and immutable (e.g. strings , numbers or tuples) , while values can be of any type.

```python
# Creating dictionaries

dict1 = {}
print(type(dict1))
```

```python
dict1 = dict()
dict1
```

```python
student = {"name":"AKSHANSH","age":20,"Grade":"A+"}
print(student)
print(type(student))
```

```python
# Single key is always used : Python keeps only the last value for a duplicate key.

student = {"name":"AKSHANSH MISHRA","age":20,"name":"A+"}
print(student)
```

```python
# Accessing elements of dictionary.

print(student['name'])
print(student['Grade'])
```

```python
# Accessing element using get method

print(student.get('Grade'))
print(student.get('last_name'))
print(student.get('last_name',"NULL")) # Giving default value to the key.
```

```python
## Modifying Dictionary Elements : Dictionary are mutable , so you can add , update or delete elements.

print(student)
```

```python
# Updating value for a key
student["age"] = 21
print(student)

# Adding a value and a key
student["address"] = "Uttar Pradesh"
print(student)

# Deleting key and value pair 
del student['name'] 
print(student)
```

```python
# Dictionary methods

# To get all the keys and values of an dictionary seperately and also in pairs

keys = student.keys()
print(keys)

values = student.values()
print(values)

items = student.items()
print(items) # List of tuples
```

```python
# Shallow Copy

student_copy = student 
print(student)
print(student_copy)

student["name"] = "AK"
print(student)
print(student_copy)

# So here in this method what happens the copy is saved in the same memory in which the student was that's why on changing in students the copy also changes.
```

```python
# To overcome this problem

student_copy1 = student.copy()

student["name"] = "AKSHANSH"
print(student_copy1)
print(student)
print(student_copy)
```

```python
## Iterating over dictionaries
## We can iterate over dictionaries for key , values , items

for keys in student.keys():
    print(keys)

for values in student.values():
    print(values)

for keys,values in student.items():
    print(f"{keys}:{values}")    
```

```python
## Nested dictionaires

students = {
    "student1" : {"name":"AKSHANSH","Package":"32 LPA"},
    "student2" : {"name":"YASH","Package":"36 LPA"}
}

print(students)
```

```python
## Access nested dictionaries elements
print(students["student1"]["name"])
print(students["student1"]["Package"])
```

```python
## Iterating over nested dictionaires

for students_id,students_info in students.items():
    print(f"{students_id}:{students_info}")
    for keys,values in students_info.items():
        print(f"{keys}:{values}")
```

```python
for value in range(8):
    print(f"{value}:{value**2}")
```

```python
## Dictionary Comprehension

squares = {x:x**2 for x in range(8)}
print(squares)
```

```python
## Conditional dictionary Comprehension

evens = {x:x**2 for x in  range(10) if x%2==0}
print(evens)
```

```python
# Use a dictionary to count the frequency of elemts in a list

numbers = [1,2,3,4,2,3,3,4,4,4]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number] = 1    
print(frequency)        
```

```python
# Merge 2 dictionaires in one

dict1 = {"a":1,"b":2}
dict2 = {"b":3 , "c":4}

merged_dict = {**dict1,**dict2}
print(merged_dict)
```
