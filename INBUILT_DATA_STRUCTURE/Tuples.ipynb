### TUPLES : Tuples are ordered collection of items that are immutable(Cannot be modified).

```python
## Creating a tuple

tpl = ()
print(type(tpl))
```

```python
# Tuple can be converted into list or vice versa.

tpl1 = tuple([1,2,3,4])
print(type(tpl1))
print(tpl1)

list1 = list((1,2,3,4))
print(type(list1))
print(list1)
```

```python
tuple2 = (1,"Hello",1.4,False)
```

```python
## Accessting tuple elements : Indexing and slicing are same in both list and tuples.

print(tuple2[0])
print(tuple2[2])
print(tuple2[-1])
```

```python
## Tuple Operations

concatenation = tuple2+tpl1
print(concatenation)
```

```python
tuple2*3
```

```python
tpl1*4
```

```python
## Immutable Nature Of Tuples

numbers = (5,6,7,7,8,8,9)

numbers[0] = 9
```

```python
## Tuple Methods

print(numbers.count(7)) # Count the occurenece of element
print(numbers.index(8)) # Returns first index of element
```

```python
## Packing and Unoacking Tuples

packed_tuple = 1,"Hey",3.14
print(packed_tuple)
```

```python
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
## Unpacking with *

num = (1,2,3,4,5,6,7)

first , *middle = num
print(first)
print(middle) # * access all the remaining elements till last.
```

```python
## nested tuples,list

lst = [[1,2,3,4],[6,7,8,9]]
lst[0][0:3]
```

```python
## Nested Tuples 

NST = ((1,2,3,4),("a","b","c"),(True,False))

print(NST[0][0:2])
print(NST[1][0:2])
```

```python
## Iterating over nested tuples

for sub_tuple in NST:
    for item in sub_tuple:
        print(item,end=" ")
    print( )    
```
